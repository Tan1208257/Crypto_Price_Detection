```python
import pandas as pd
from src.models import get_models
from src.evaluation import evaluate

df = pd.read_csv("../data/processed/merged_features.csv")
df["timestamp"] = pd.to_datetime(df["timestamp"])
```

```python
features = [c for c in df.columns if c not in ["timestamp", "target", "symbol"]]

train = df[df["timestamp"] < "2024-01-01"]
test = df[df["timestamp"] >= "2024-01-01"]

X_train, y_train = train[features], train["target"]
X_test, y_test = test[features], test["target"]
```

```python
models = get_models()

for name, model in models.items():
    model.fit(X_train, y_train)
    print(name, evaluate(model, X_test, y_test))
```
